```python
import csv
from pathlib import Path
import matplotlib.pyplot as plt

files = sorted(Path(".").glob("AFM_fit_points_curve_*.csv"))

if not files:
    print("No curve CSV files found. Check the notebook's current folder.")

for file in files:
    with file.open(newline="") as f:
        rows = list(csv.DictReader(f))

    if not rows:
        continue

    x = [float(row["x_nm"]) for row in rows]
    y = [float(row["measured_y_nN"]) for row in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(x, y, ".", markersize=3, label="Measured")

    for column, label in [
        ("linear_fit_y_nN", "Linear fit"),
        ("elasticity_fit_y_nN", "Elasticity fit"),
    ]:
        points = sorted(
            (float(row["x_nm"]), float(row[column]))
            for row in rows
            if row[column].strip()
        )

        if points:
            fit_x, fit_y = zip(*points)
            plt.plot(fit_x, fit_y, linewidth=2, label=label)

    plt.title(f"Curve {rows[0]['curve_number']}")
    plt.xlabel("X (nm)")
    plt.ylabel("Force (nN)")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
